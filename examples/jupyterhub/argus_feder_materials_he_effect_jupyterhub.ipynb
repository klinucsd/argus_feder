## The question

A burning plasma makes helium as well as consuming hydrogen isotopes, so the
tungsten facing it is helium-conditioned tungsten: bubbles form a short distance
below the surface long before the wall reaches end of life. Whether that
conditioned surface then holds **more or less fuel** is a first-order question
for a reactor, because deuterium and tritium held in the wall are a safety limit
and an economic one.

The delivery this notebook works from was built to ask exactly that. Tungsten
coupons were pre-exposed to helium plasma in the PISCES-A linear device at two
temperatures, mounted on the DiMES probe alongside coupons that were not, and
exposed together to real DIII-D discharges. Retention was then measured
afterwards by several methods.

The question is asked plainly first, and then pressed on the points where a
single number could hide the answer: which measurement was used, which plasma
the coupon saw, and where in the depth the deuterium actually sits.
